**Problem 6.** *Does the cap pay for itself?* The support plan from problem 4 costs 1 credit
a night and covers whatever a night's bill runs over 25. So what it saves you on one night is

$$W = \max(X - 25,\ 0)$$

which is 75 credits on a 100-credit night and 0 on every other night.

`norm` is already imported: `norm.cdf(z)` returns $\mathbb{P}(Z < z)$ for a standard normal.

```python
# From this point on, it doesn't specify "by hand" so I had AI write code.
```

Part a. Compute $\mathbb{E}[W]$ and $\mathrm{SD}(W)$ from the five outcomes of $X$.

```python
# W = max(X - 25, 0): the outcomes 1, 2, 5, 20, 100 give savings 0, 0, 0, 0, 75
EW  = 0*0.50 + 0*0.25 + 0*0.15 + 0*0.08 + 75*0.02
EW2 = 0**2*0.50 + 0**2*0.25 + 0**2*0.15 + 0**2*0.08 + 75**2*0.02
VarW = EW2 - EW**2
SDW = np.sqrt(VarW)
print("E[W] =", EW, " E[W^2] =", EW2, " Var(W) =", VarW, " SD(W) =", SDW)
```

Part b. Over 30 nights the plan costs 30 credits and saves a total of $S = W_1 + \cdots + W_{30}$. Give $\mathbb{E}[S]$ and $\mathrm{SD}(S)$, then use the central limit theorem to approximate $\mathbb{P}(S > 30)$, the chance the plan pays for itself over the month.

```python
# S = W1 + ... + W30, 30 independent nights
n = 30
ES = n * EW                # 30 * 1.5
VarS = n * VarW            # 30 * 110.25
SDS = np.sqrt(VarS)
z = (30 - ES) / SDS        # plan costs 30 credits
p_pays = 1 - norm.cdf(z)   # P(S > 30)
print("E[S] =", ES, " Var(S) =", VarS, " SD(S) =", SDS, " z =", z, " P(S > 30) =", p_pays)
```

Part c. Do the same for 100 nights, where the plan costs 100 credits. Report $\mathbb{P}(S > 100)$ and say in one sentence why it is larger than your answer to part b, given that the plan saves 1.50 a night either way.

```python
# Same thing over 100 nights
n = 100
ES100 = n * EW               # 100 * 1.5
VarS100 = n * VarW           # 100 * 110.25
SDS100 = np.sqrt(VarS100)
z100 = (100 - ES100) / SDS100   # plan costs 100 credits
p_pays100 = 1 - norm.cdf(z100)  # P(S > 100)
print("E[S] =", ES100, " Var(S) =", VarS100, " SD(S) =", SDS100, " z =", z100, " P(S > 100) =", p_pays100)
```

_Your answer:_ Over 100 nights, Pr(S > 100) is about 0.683, compared with about 0.603 over 30 nights. The expected surplus (1.50 saved minus 1 paid per night) grows in proportion to the number of nights, but the SD of the total savings grows only like the square root of the number of nights, so with more nights the break-even point sits further below the expected savings in SD units (z goes from -0.26 to -0.48), and the chance the plan pays for itself goes up.

Part d. Now run the same approximation for a **single** night: treat $W$ as normal with the mean and standard deviation from part a, and approximate $\mathbb{P}(W > 1)$. Then compute the exact $\mathbb{P}(W > 1)$ straight from the table in problem 3. Report both.

```python
# One night: normal approximation vs exact
z1 = (1 - EW) / SDW              # (1 - 1.5) / 10.5
p_normal = 1 - norm.cdf(z1)      # approx P(W > 1)
p_exact = 0.02                   # W > 1 only when X = 100, which has probability 0.02
print("z =", z1, " normal approx P(W > 1) =", p_normal, " exact P(W > 1) =", p_exact)
```

Part e. Your two numbers in part d are not close. Explain what went wrong, and state the condition the central limit theorem needs that one night does not meet.

_Your answer:_ The normal approximation says a single night has about a 52% chance of saving more than 1 credit, but the exact answer is only 2%. One night's savings W is nothing like a normal distribution: it is 0 on 98% of nights and 75 on 2% of nights, so it is extremely right-skewed with just two possible values, and a bell curve centered at 1.5 with SD 10.5 puts about half its area above 1 even though W almost never lands there. The central limit theorem is about sums or averages of many independent observations: it needs a large enough n for the sum to look normal.

**Problem 7.** *Probabilities estimated from data.* Back to the 600 moving jobs. Define two
events:

- $A$: the job took more than 12 hours
- $B$: the job moved more than 800 cubic feet

Estimate each probability below by counting the rows where it holds and dividing by 600. In pandas,
`(jobs.hours > 12).mean()` does the counting and the dividing in one step.

Part a. Estimate $\mathbb{P}(A)$, $\mathbb{P}(B)$ and $\mathbb{P}(A \text{ and } B)$.

```python
# A: more than 12 hours, B: more than 800 cubic feet
A = jobs.hours > 12
B = jobs.volume_cuft > 800
pA = A.mean()          # count of A rows / 600
pB = B.mean()          # count of B rows / 600
pAB = (A & B).mean()   # count of rows with both / 600
print("counts: A =", A.sum(), " B =", B.sum(), " A and B =", (A & B).sum())
print("P(A) =", pA, " P(B) =", pB, " P(A and B) =", pAB)
```

Part b. Estimate $\mathbb{P}(A \mid B)$ and $\mathbb{P}(B \mid A)$.

```python
pA_given_B = pAB / pB   # P(A and B) / P(B)
pB_given_A = pAB / pA   # P(A and B) / P(A)
print("P(A | B) =", pA_given_B, " P(B | A) =", pB_given_A)
```

Part c. Check whether $A$ and $B$ are independent. Report the two numbers you compared, and say what you conclude and why it is the answer you would have expected here.

```python
# If independent, P(A | B) would equal P(A), and P(A and B) would equal P(A) * P(B)
print("P(A | B) =", pA_given_B, " vs  P(A) =", pA)
print("P(A and B) =", pAB, " vs  P(A) * P(B) =", pA * pB)
```

_Your answer:_ I compared Pr(A | B) = 0.781 with Pr(A) = 0.278 (and, as a second check, Pr(A and B) = 0.255 with Pr(A)P(B) = 0.091). These are nowhere near equal: knowing a job moved more than 800 cubic feet nearly triples the chance it took more than 12 hours, so A and B are clearly not independent. This is what we would expect, because moving more stuff takes more time to pack, load, carry and unload, so big moves and long jobs naturally go together.
